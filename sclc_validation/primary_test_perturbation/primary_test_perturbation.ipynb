# Primary donor-held-out SCLC perturbation analysis

This notebook reviews only test-cell perturbations with training-only Geneformer V2 reference centroids. It must not be used to pool training, evaluation, or whole-cohort results into the primary estimate.

```python
from pathlib import Path
import pandas as pd
HERE = Path.cwd()
TABLES = HERE / 'tables'
summary = pd.read_csv(TABLES / 'primary_arm_summary.csv')
audit = pd.read_csv(TABLES / 'coverage_audit.csv')
hits = pd.read_csv(TABLES / 'primary_concordant_hits.csv')
summary
```

## Completeness checks

All six directional comparisons are required for each arm. Missing files are pending computation, not zero-effect findings.

```python
assert set(summary.arm) <= {'delete', 'overexpress'}
assert len(audit) == 12
print(audit[['arm', 'comparison', 'status', 'rows']].to_string(index=False))
print('Complete tables:', (audit.status == 'ok').sum(), '/', len(audit))
```

## Primary concordance

A hit requires FDR < 0.05 in both arms, opposite-signed shifts, and at least 25 deletion detections. These filters do not replace donor-level consistency or biological validation.

```python
hits.sort_values(['comparison', 'abs_delete_shift'], ascending=[True, False]).head(50)
```

## Interpretation boundary

Geneformer deletion and overexpression are rank-encoding interventions. They measure model representation sensitivity and should be described as candidate prioritization, not as validated gene knockout effects.
